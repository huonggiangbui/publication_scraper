```python
import pandas as pd
from transformers import pipeline

# classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")
classifier = pipeline(
    "zero-shot-classification", model="typeform/distilbert-base-uncased-mnli"
)
```

```python
candidate_labels = [
    "Relevant to the systematic review on responsible or ethical use of LLMs in education.",
    "Not relevant to the review.",
]
```

```python
df = pd.read_csv("sample-data/cleaned-results-sch-full.csv")

if "Abstract" not in df.columns:
  raise ValueError("CSV file must contain an 'abstract' column")
```

```python
predicted_labels = []
predicted_scores = []

for abstract in df["Abstract"]:
    if isinstance(abstract, str) and abstract.strip():
        result = classifier(abstract, candidate_labels)
        predicted_labels.append(result["labels"][0])
        predicted_scores.append(result["scores"][0])
    else:
        # Handle empty or invalid abstracts
        predicted_labels.append("No abstract")
        predicted_scores.append(None)

# Add results to dataframe
df["predicted_label"] = predicted_labels
df["predicted_score"] = predicted_scores

# Save to new CSV
df.to_csv("sample-data/classified_abstracts.csv", index=False)

print("Classification complete. Results saved to 'classified_abstracts.csv'.")
```
